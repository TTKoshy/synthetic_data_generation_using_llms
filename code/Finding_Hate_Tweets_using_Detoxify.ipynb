# Objective:
This notebook was created to identify additional hate tweets from topics that contained fewer than 20 hate tweets in our manually labeled Twitter dataset. The output from the Detoxify model will be manually reviewed, and the hate tweets identified will be added to the final dataset.

```python
import pandas as pd
from detoxify import Detoxify
from tqdm import tqdm
```

```python
THRESH = {
    "toxicity": 0.90,
    "severe_toxicity": 0.80,
    "identity_attack": 0.80,
    "insult": 0.90,
    "threat": 0.70,
}
KEEP_IF_ANY = True   # True = OR, False = AND
BATCH_SIZE = 128
```

```python
model = Detoxify("unbiased")
scores = {k: [] for k in ["toxicity","severe_toxicity","obscene","threat","insult","identity_attack","sexual_explicit"]}
```

# Identify high priority hate tweets from 2000 sample tweets in Topic 11

```python
IN_CSV = "topic11_sample_2000.csv"
OUT_FILE = "high_prob_detoxify_topic11_part2.csv"
```

```python
df = pd.read_csv(IN_CSV).dropna(subset=["clean_text"])
df["text"] = df["clean_text"].astype(str)
```

```python
for i in tqdm(range(0, len(df), BATCH_SIZE)):
    batch = df["text"].iloc[i:i+BATCH_SIZE].tolist()
    out = model.predict(batch)
    for k in scores:
        scores[k].extend(out.get(k, [None]*len(batch)))

for k in scores:
    df[k] = scores[k]

def is_high(row):
    hits = []
    for attr, thr in THRESH.items():
        val = row.get(attr, 0.0)
        hits.append(val is not None and val >= thr)
    return any(hits) if KEEP_IF_ANY else all(hits)

high = df[df.apply(is_high, axis=1)].copy()
high = high.sort_values(list(THRESH.keys()), ascending=False)

high.to_csv(OUT_FILE, index=False)
print(f"Saved {len(high)} high-probability tweets to {OUT_FILE}")
```
